## 生成周报框架

```python
from pptx import Presentation
from pptx.util import Cm  # 也可用 Inches
import requests
from PIL import Image
from io import BytesIO
import os

# 图片URL列表
urls = [
    'https://www.cpc.ncep.noaa.gov/products/JAWF_Monitoring/SEAsia/daily/p.7day.figb.gif',
    'https://www.cpc.ncep.noaa.gov/products/JAWF_Monitoring/GFS/SEAsia_curr.p.gfs1b.gif',
    'https://www.cpc.ncep.noaa.gov/products/JAWF_Monitoring/GFS/SEAsia_curr.p.gfs2b.gif',
    'https://www.cpc.ncep.noaa.gov/products/JAWF_Monitoring/SEAsia/daily/p.7day.figa.gif',
    'https://www.cpc.ncep.noaa.gov/products/JAWF_Monitoring/GFS/SEAsia_curr.p.gfs1a.gif',
    'https://www.cpc.ncep.noaa.gov/products/JAWF_Monitoring/GFS/SEAsia_curr.p.gfs2a.gif'
]
noaa_imgs = []

for url in urls:
    response = requests.get(url)
    if response.status_code == 200:
        img = Image.open(BytesIO(response.content))
        img_converted = img.convert("RGB")

        # 保存为临时文件
        temp_img_path = f"temp_{urls.index(url)}.jpg"  # 使用图片的URL索引来生成唯一文件名
        img_converted.save(temp_img_path)

        # 将图片路径保存到列表
        noaa_imgs.append(temp_img_path)
    else:
        print(f'Download fail: "{url}"')



ppt_path = '天气周报.pptx'
save_path = '天气周报2.pptx'
img_path_list = ['./../diagram/grid/wr1.jpg',
                 './../diagram/grid/wr2.jpg',
                 './../diagram/grid/wr3.jpg',
                 './../diagram/grid/wr4.jpg']

prs = Presentation(ppt_path)

slide = prs.slides[6]
high = Cm(11)

i = 0
for noaa_top in [ Cm(5.00), Cm(20.0)]:
    for noaa_left in [ Cm(1.45), Cm(12.8), Cm(23.15)]:
        pic_shape = slide.shapes.add_picture(noaa_imgs[i], noaa_left, noaa_top, height=Cm(11))
        i = i + 1

for idx, img_path in enumerate(img_path_list):
    slide = prs.slides[ idx + 7 ]
    left = Cm(2.58)
    top = Cm(3.49)
    width = Cm(38.9)   # 指定 width，不给 height 会保持图片原始纵横比
    pic_shape = slide.shapes.add_picture(img_path, left, top, width=width)


prs.save(save_path)
print("已保存：", save_path)

# 删除临时图片文件
for temp_img in noaa_imgs:
    os.remove(temp_img)
```
